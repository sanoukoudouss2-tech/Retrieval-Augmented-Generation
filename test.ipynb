```python
from src.extraction import extraction
from src.contexte import contexte
from src.decoupage import coupure
from src.proximite import proximite
from src.vectorisation import vectorisation
```

```python
texte = extraction("./le_petit_prince.pdf")
print(texte)
```

```python
decoupe = coupure(texte,100,15)
print(len(decoupe))
```

```python
base_vec = vectorisation(decoupe)
print(base_vec.shape)
```

```python
Q_1 = "Qui est le renard?"
Q_2 = "Pourquoi le renard demande t'il au petit prince de l'apprivoiser?"
Q_3 = "Que signifie pour le renard le fait d'être apprivoisé?"
Q_4 = "Pourquoi le petit prince finit il par considérer sa rose comme uniquealors qu'il existe des milliers de roses?"
```

```python
indices_1 = proximite(Q_1,base_vec)
indices_2 = proximite(Q_2,base_vec)
indices_3 = proximite(Q_3,base_vec)
indices_4 = proximite(Q_4,base_vec)

```

```python
ctxte_1 = contexte(indices_1,decoupe)
ctxte_2 = contexte(indices_2,decoupe)
ctxte_3 = contexte(indices_3,decoupe)
ctxte_4 = contexte(indices_4,decoupe)
```

```python
## API
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
from groq import Groq
from src.prompt import prompt 
client = Groq(api_key=api_key)

```

```python
requette_1 = prompt(Q_1,ctxte_1)
response = client.chat.completions.create(model="openai/gpt-oss-120b",
                                          messages= requette_1)

print(response.choices[0].message.content)
```

```python
requette_2 = prompt(Q_2,ctxte_2)
response = client.chat.completions.create(model="openai/gpt-oss-120b",
                                          messages= requette_2)

print(response.choices[0].message.content)
```

```python
requette_3 = prompt(Q_3,ctxte_3)
response = client.chat.completions.create(model="openai/gpt-oss-120b",
                                          messages= requette_3)

print(response.choices[0].message.content)
```

```python
requette_4 = prompt(Q_4,ctxte_4)
response = client.chat.completions.create(model="openai/gpt-oss-120b",
                                          messages= requette_4)

print(response.choices[0].message.content)
```
